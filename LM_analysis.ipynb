```python
from transformers import pipeline
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fill_mask = pipeline("fill-mask", model="bert-base-uncased")
result = fill_mask("Artificial Intelligence is a [MASK] field.")
for r in result:
    print(f"{r['sequence']} (score: {r['score']:.4f})")
```

```python
generator = pipeline("text-generation", model="gpt2")
generated = generator("Natural language processing enables", max_length=50, num_return_sequences=1)
print(generated[0]['generated_text'])
```

```python
prompts = ["AI is a [MASK] field.", "Data science is a [MASK] domain."]
scores = [fill_mask(prompt)[0]['score'] for prompt in prompts]

sns.barplot(x=["AI", "Data Science"], y=scores)
plt.title("Confidence for Top Prediction")
plt.ylabel("Score")
plt.show()
```
